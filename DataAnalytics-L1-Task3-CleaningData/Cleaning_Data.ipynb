```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("retail_store_sales.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df.describe(include='all')
```

```python
missing = df.isnull().sum()

missing_percent = (df.isnull().sum() / len(df) * 100).round(2)

quality_missing = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_percent
})

quality_missing
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicate_count)
```

```python
df = df.drop_duplicates().copy()

print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
text_columns = df.select_dtypes(include=['object', 'string']).columns

print("Text columns:")
print(text_columns.tolist())
```

```python
for col in text_columns:
    df[col] = df[col].astype('string').str.strip()

print("Text formatting standardized.")
```

```python
print(df.dtypes)
```

```python
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

print(df.dtypes)
```

```python
print("Before:")
print(df.dtypes)

if 'Transaction Date' in df.columns:
    df['Transaction Date'] = pd.to_datetime(
        df['Transaction Date'], errors='coerce'
    )

print("\nAfter:")
print(df.dtypes)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numeric columns:")
print(numeric_columns.tolist())
```

```python
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print(df.dtypes)
```

```python
print(df.columns.tolist())
print("\nMissing values:")
print(df.isnull().sum())
print("\nData types:")
print(df.dtypes)
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0]

print("Missing values:")
print(missing)
```

```python
missing_report = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing %': (df.isnull().sum() / len(df) * 100).round(2)
})

missing_report = missing_report[missing_report['Missing Count'] > 0]

missing_report
```

```python
print(df.columns.tolist())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Numeric columns → median
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Text columns → mode
text_cols = df.select_dtypes(include=['object', 'string']).columns

for col in text_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
total_missing = df.isnull().sum().sum()

print("Total missing values:", total_missing)
```

```python
outlier_report = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_count = ((df[col] < lower) | (df[col] > upper)).sum()
    outlier_report[col] = outlier_count

outlier_report
```

```python
outlier_report
```

```python
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower=lower, upper=upper)

print("Outliers capped using IQR method.")
```

```python
outlier_after = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_after[col] = ((df[col] < lower) | (df[col] > upper)).sum()

print(outlier_after)
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Total rows:", len(df))
print("Total missing values:", df.isnull().sum().sum())
```
