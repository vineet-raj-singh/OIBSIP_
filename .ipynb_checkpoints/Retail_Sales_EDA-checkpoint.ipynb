```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Date'].head()
```

```python
monthly_sales = df.groupby(df['Date'].dt.to_period('M'))['Total Amount'].sum()

monthly_sales
```

monthly_sales = df.groupby(df['Date'].dt.to_period('M'))['Total Amount'].sum()

monthly_sales

### Observation
The monthly sales trend shows how sales revenue changes over time. Peaks indicate months with higher sales, while lower points indicate periods of reduced sales.

```python
monthly_sales.plot(kind='line', figsize=(12,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
quarterly_sales = df.groupby(df['Date'].dt.to_period('Q'))['Total Amount'].sum()

quarterly_sales
```

### Observation
The quarterly sales analysis helps identify broader sales patterns and shows which quarters performed better or worse.

```python
quarterly_sales.plot(kind='line', marker='o', figsize=(10,5))

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
bins = [0, 18, 25, 35, 45, 55, 100]
labels = ['Under 18', '18-25', '26-35', '36-45', '46-55', '56+']

df['Age Group'] = pd.cut(
    df['Age'],
    bins=bins,
    labels=labels
)

df['Age Group'].value_counts().sort_index()
```

### Observation
The age-group distribution shows which customer age groups contribute the largest share of the customer base. This can help the business target its marketing efforts more effectively.

```python
df['Age Group'].value_counts().sort_index().plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Customer Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```

```python
gender_counts = df['Gender'].value_counts()

gender_counts
```

### Observation
The gender distribution shows the proportion of male and female customers in the dataset. This can help understand the overall customer profile.

```python
gender_counts.plot(
    kind='bar',
    figsize=(7,5)
)

plt.title('Customer Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```

```python
category_sales = df.groupby('Product Category')['Total Amount'].sum()

category_sales
```

### Observation
The revenue comparison across product categories shows which categories generate the highest sales. Higher-performing categories can be prioritized for inventory and promotional campaigns.

```python
category_sales.sort_values(ascending=False).plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=0)
plt.show()
```

```python
category_quantity = df.groupby('Product Category')['Quantity'].sum()

category_quantity
```

```python
numeric_data = df.select_dtypes(include='number')

correlation = numeric_data.corr()

correlation
```

### Observation
The correlation heatmap shows the strength and direction of relationships between numerical variables. Stronger correlations indicate variables that tend to change together.

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```

### Observation
The scatter plot helps examine whether customer age has a visible relationship with spending. The distribution of points can reveal whether spending behaviour differs across age groups.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Age',
    y='Total Amount',
    hue='Gender'
)

plt.title('Age vs Total Amount')
plt.xlabel('Age')
plt.ylabel('Total Amount')
plt.show()
```
